## Section 3.3.1: Necessary and sufficient conditions of local minimizers

To find the necessary and sufficient conditions of local maximizers. we can use functions and their derivative functions to understand where the critical points with maxima and minima are at. Below, we demonstrate an example of g(x), g'(x), and g''(x) to show where critical points can be in the function.


```python
# Functon of g(x) = sin(x)^2
def g(x):
    return (np.sin(x))**2
# Function of g'(x) = 2sin(x)cos(x)
def gp(x):
    return 2 * np.sin(x) * np.cos(x)
# Function of g''(x) = 2cos^2(x) - 2sin^2(x)
def gpp(x):
    return 2 * ((np.cos(x))**2 - (np.sin(x))**2)
# Critical Points range [-pi, pi]
critical_points = np.array([-np.pi, 0, np.pi])
# Find all critical points.
for c in critical_points:
    print(f"g({c}) = {g(c)}, gp({c}) = {gp(c)}, gpp({c}) = {gpp(c)}")
# Graph the critical points of g(x) from -pi to pi.
x = np.linspace(-np.pi, np.pi, 300)
plt.plot(x, g(x), label="g(x)")
plt.scatter(critical_points, g(critical_points), color="tab:red", label="critical points")
plt.legend()
plt.title("Critical points of g(x)")
plt.show()
```

## 3.3.2: Convexity and global minimizers

For functions, we can also analyze the convexity and global minimizers to understand whether a certain interval of a function may be curved up or down. As shown in this example, we can understand how a convex function may show global minimizers to understand graph behavior.

```python
a = np.random.randint(0,9)
def h(x):
    return (x - 2)**a + 1
print("Information Regarding h(x) and h'(x):")
print(f"h(x) = (x - 2)^{a} + 1")
print(f"h'(x) = {a} * (x - 2)^{a - 1}")

x = np.linspace(0, 4, 300)
plt.plot(x, h(x))
plt.axvline(2, color="tab:red", linestyle="--", label="global minimizer")
plt.scatter(2, h(2), color="tab:red", label="global minimizer")
plt.title("Convex Function h(x) with a global minimizer")
plt.legend()
plt.show()
```

## 3.3.3: Gradient descent

Gradient descent is a mathematical method used to find the lowest possible error in a function via iteration. Given learning rate, path and current position, we can understand the exact point where a function minimizes its error.

We demonstrate this using a linear function, understanding how learning rate may affect the final result.

```python
def grad_g(x):
    return ((1 / 2) * x) - 4
# Set Learning Rate, current value, and the path value
learning_rate = 0.16
current = -3.4
path = [current]
random_num = np.random.randint(50,100)
for _ in range(random_num):
    current = current - learning_rate * grad_g(current)
    path.append(current)

path = np.array(path)
print("Iterations =", random_num)
print("Final x =", path[-1])
print("Final objective =", g(path[-1]))

plt.plot(path, marker="o", label="Path")
plt.axhline(0, color="tab:red", linestyle="--", label="y = 0")
plt.xlabel("Iteration")
plt.ylabel("X-value")
plt.title("Gradient Descent Iteration")
plt.legend()
plt.show()
```
