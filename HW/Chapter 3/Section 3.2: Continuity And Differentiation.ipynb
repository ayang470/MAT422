## Section 3.2.1: Limits and Continuity
A limit is defined in a function f, where for any ϵ > 0, the function output f(x) approaches arbitrarly close to a limit L within a certain range δ > 0. Limits measure continuity, where a function f is continuous at a point a if the limit f(x) as x approaches a exists and equals function value f(a).

In this example, we demonstrate two functions f(x) = 1/x and g(x) = sin(x) and use numpy determine the continuity of this function. Additionally, we can demonstrate an asymptote discontinuity using matplotlib for f(x).

```python
h = np.array([1, 0.5, 0.1, 0.001, 0.0001])
values = 1 / h
values2 = np.sin(h)
print("FUNCTION 1: f(x) = 1/x")
print("h values:", h)
print("1/h values:", values)
print("These values do not show continuity of the function f(x).")
print("FUNCTION 2: g(x) = sin(x)")
print("h values:", h)
print("sin(h) values:", values2)
print("These values show continuity of the function g(x).")

x = np.linspace(-10, 10, 100)
y = 1/x
plt.plot(x,y)
plt.axvline(x=0, color='r', linestyle='--')
plt.axhline(y=0, color='r', linestyle='--')
plt.title("Example of Asymptote Discontinuity at x = 0 and y = 0")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.show()
```

## Section 3.2.2: Derivatives

For a function, the derivative measures how sensitive a function value changes with respect to another variable. Using derivatives, we can understand how exactly a function changes at different points to determine increasing or decreasing properties.

This example demonstrates how derivatives are measured at a point to show differentiation and properties of a quadratic function.

```python
# Define a function x^2 + 3x + 2.
def f(x):
  return x ** 2 + 3 * x + 2
# Define Points, Steps, Approx Derivatives, and Exact Derivatives
point = 3
steps = np.array([1e-1, 1e-2, 1e-3, 1e-4])
approx_derivatives = (f(point + steps) - f(point - steps)) / (2 * steps)
exact_derivative = 2 * point + 3
# Print out Necessary derivatives and errors.
print("Approximate Derivatives:", approx_derivatives)
print("Exact Derivative:", exact_derivative)
print("Errors:", np.abs(approx_derivatives - exact_derivative))
```

## Section 1.2.3: Taylor's Theorem


Taylor's Theorem approximates a differentiable function using a given point from a polynominal. This theorem approximates the mean value theorem, as it multiplies an order of the function derivative by powers of the difference, scaled by factorials with remainders.

In this example, we can approximate a function f using taylor approximation to show the accuracy between exact and approximation values.

```python
x = np.linspace(-2, 2, 300)
taylor_exp = 1 + x + x**2/2 + x**3/6 + x**4/24 + x**5/120
plt.plot(x, np.exp(x), label='exact')
plt.plot(x, taylor_exp, "--", label='Fifth-degree Taylor polynomial at 0')
plt.xlabel("x")
plt.ylabel("value")
plt.title("Taylor's 5th Polynomial Approximation")
plt.legend()
plt.show()

print("Max error on [-2, 2] =", np.max(np.abs(np.exp(x[np.abs(x) <= 1] - taylor_exp[np.abs(x) <= 1]))))
```
